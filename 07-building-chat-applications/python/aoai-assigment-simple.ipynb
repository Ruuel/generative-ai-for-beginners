```python
import os
from openai import AzureOpenAI

API_KEY = os.getenv("AZURE_OPENAI_API_KEY", "").strip()
assert API_KEY, "ERROR: Azure OpenAI Key is missing"

RESOURCE_ENDPOINT = os.getenv("AZURE_OPENAI_ENDPOINT", "").strip()
assert RESOURCE_ENDPOINT, "ERROR: Azure OpenAI Endpoint is missing"
assert "openai.azure.com" in RESOURCE_ENDPOINT.lower(), "ERROR: Azure OpenAI Endpoint should be in the form: \n\n\t<your unique endpoint identifier>.openai.azure.com"

client = AzureOpenAI(
    api_key=API_KEY,
    azure_endpoint=RESOURCE_ENDPOINT,
    # Dated api_version strings get retired roughly a year after a newer one
    # ships - see https://learn.microsoft.com/azure/ai-foundry/openai/api-version-lifecycle
    # for the current value if this one stops working.
    api_version=os.getenv("AZURE_OPENAI_API_VERSION", "2024-10-21").strip(),
)

deployment = "gpt-4o-mini" # replace with your deployment name (gpt-35-turbo has been retired by Azure OpenAI)
```

```python
# Create your first prompt
text_prompt = " My foot hurts, what can be wrong?"

response = client.chat.completions.create(
  model=deployment,
  messages = [
      {"role":"system", "content":"I'm a doctor, specialist on surgery"},
      {"role":"user","content":text_prompt},])


response.choices[0].message.content
```
